# Synthetic Data Generator

This project generates realistic synthetic data for analytics practice, SQL testing, and portfolio projects.

Features:
- Prefixed readable Entity identifier
- Realistic product category launch timeline

## Imports

```python
from faker import Faker
import pandas as pd
import random
import datetime
```

## Generate Synthetic Dataset

```python
def create_synthetic_data():

    fake = Faker('en_TH')
    # `seed 92` is initialized to guarantee deterministic record generation, ensuring permanent identifier consistency for entity-relationship diagrams (ERDs) and database joins.
    random.seed(92)
    fake.seed_instance(92)

    Category_ID = []
    Category_Name = [
        'Mattress',
        'Bedsheet',
        'Pillow',
        'Bolster',
        'Blanket'
    ]
    Created_At = []

    for name in Category_Name:
        # Distinct entity prefixes, such as `ProCat_` for Product_Category and `ORD_` for orders, are implemented to prevent identifier collision during table joins. 
        prefix_ID = "PROCAT_"
        Category_ID.append(prefix_ID + fake.uuid4()[:4])

        # Simulated product-line expansion where Mattress and Bedsheet launch first, followed by Pillow, Bolster, and Blanket in later months.
        if name == 'Mattress' or name == 'Bedsheet':
            Created_At.append(datetime.date(2025, 1, 1))
        else:
            Created_At.append(datetime.date(2025, 4, 1))

    Product_Category = pd.DataFrame({
        'Category_ID': Category_ID,
        'Category_Name': Category_Name,
        'Created_At': Created_At
    })
    return Product_Category
```

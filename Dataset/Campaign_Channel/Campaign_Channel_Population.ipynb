# Synthetic Data Generator

This project generates realistic synthetic data for analytics practice, SQL testing, and portfolio projects.


## Imports

```python
from faker import Faker
import pandas as pd
import random
```

## Generate Synthetic Data

```python
def create_synthetic_data(n_rows):
    fake = Faker('en_TH')
    SEED = 92
    random.seed(SEED)
    fake.seed_instance(SEED)

    Campaign_ID = []
    Channel_ID = []

    for i in range(n_rows):
        Campaign_ID.append(fake.unique.uuid4()[:])
        Channel_ID.append(fake.unique.uuid4()[:])

    data = pd.DataFrame({
        'Campaign_ID': Campaign_ID,
        'Channel_ID': Channel_ID
    })
    return data
        

```

```python
campaign_channel_data = create_synthetic_data(5)
campaign_channel_data
```
